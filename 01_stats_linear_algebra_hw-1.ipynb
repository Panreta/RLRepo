## Linear algebra (40pts)
- 1.1. Write this fixed point iteration in matrix form
$$
\begin{aligned}
x_1 &= 0.2 x_2 + 0.7 x_3 \\
x_2 &= 0.1 x_1 + 0.8 x_3 \\
x_3 &= 0.2 x_1 + 0.7 x_2
\end{aligned}
$$
- 1.2. Is the matrix sub-stochastic?
- 1.3. Let $x_1 = 1, x_2 = 1, x_3 = 1$, do 2 iteration by hand using matrix-vector product, and show results


# Solution

```python
# 1.1
import numpy as np

def fix_point(A, x0, tol=1e-6, max_iter=1000):
    x = x0
    for _ in range(max_iter):
        x_new = A @ x
        if np.linalg.norm(x_new - x) < tol:
            return x_new
        x = x_new
    # raise ValueError("Fixed point iteration did not converge within the maximum number of iterations.")
    return x_new

# 1.2

A = np.array([
    [0.0, 0.2, 0.7],
    [0.1, 0.0, 0.8],
    [0.2, 0.7, 0.0]
])

row_sums = A.sum(axis=1)
is_substochastic = np.all(A >= 0) and np.all(row_sums <= 1)
# the result is [0.9 0.9 0.9], and for each entry it's between 0 and 1, so it's substochastic.

# 1.3
x0 = np.array([1.0, 1.0, 1.0])
fixed_point = fix_point(A, x0,tol=1e-6, max_iter=2)
print("Fixed point:", fixed_point)
# The answer is  [0.81 0.81 0.81].
```

## Linear algebra (40pts)
- 2.1 Write a function that takes a square matrix $A$, an input vector $x$, and a vector $b$, and an integer $maxit$ (default =10) and return the result of $maxit$ iterations of
$$ x \leftarrow A x + b$$
- 2.2 Now take A as the matrix in the previous problem. $x = [1,2,3]$, $b = [3,2,1]$, and do 10, 100, 1000 iterations. Does the iteration process converge? To what?
- 2.4 Now instead of $A$, use $1.2*A$, do 10, 100, 1000 iterations, does the iteration converge?
- 2.5 Relate what you see to the contractive mapping theorem 

# Solution

```python
# 2.1
import numpy as np

def iterate(A, x, b, maxit=10):
    x = np.array(x, dtype=float)
    b = np.array(b, dtype=float)
    for _ in range(maxit):
        x = A @ x + b
    return x

# 2.2
A = np.array([
    [0.0, 0.2, 0.7],
    [0.1, 0.0, 0.8],
    [0.2, 0.7, 0.0]
])
x0, b = [1.0, 2.0, 3.0], [3.0, 2.0, 1.0]

result10 = iterate(A, x0, b, 10)    
result100 = iterate(A, x0, b, 100)   
result1000 = iterate(A, x0, b, 1000)
gap_10_100   = np.linalg.norm(result100  - result10)
gap_100_1000 = np.linalg.norm(result1000 - result100)

if gap_100_1000 < 1e-3 and gap_100_1000 < gap_10_100:
    print(f"Converge! Gap shrank from {gap_10_100:.4f} to {gap_100_1000:.2e}")
    print(f"Fixed point ≈ {result1000}")

# 2.4
A2 = 1.2 * A
result10 = iterate(A2, x0, b, 10)    
result100 = iterate(A2, x0, b, 100)   
result1000 = iterate(A2, x0, b, 1000)
# Print find the answer diverged.
gap_10_100   = np.linalg.norm(result100  - result10)
gap_100_1000 = np.linalg.norm(result1000 - result100)
if gap_100_1000 < 1e-3 and gap_100_1000 < gap_10_100:
    print(f"Converge! Gap shrank from {gap_10_100:.4f} to {gap_100_1000:.2e}")
    print(f"Fixed point ≈ {result1000}")
else:
    print("Diverge")
```
